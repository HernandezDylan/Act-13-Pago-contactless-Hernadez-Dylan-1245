#### Pago contactless

```python
class Pago_Contactless:

    def __init__(self, numero_tarjeta, saldo_disponible,
                 nombre_usuario, limite_pago, estado_tarjeta):

        self.numero_tarjeta = numero_tarjeta
        self.saldo_disponible = saldo_disponible
        self.nombre_usuario = nombre_usuario
        self.limite_pago = limite_pago
        self.estado_tarjeta = estado_tarjeta

    # Función para realizar pago
    def realizar_pago(self):

        if self.estado_tarjeta == "Activa":

            cantidad = float(input("Ingrese cantidad a pagar: "))

            if cantidad <= self.saldo_disponible:

                if cantidad <= self.limite_pago:

                    self.saldo_disponible -= cantidad
                    print("Pago realizado correctamente")
                    print("Saldo restante:", self.saldo_disponible)

                else:
                    print("El pago supera el limite permitido")

            else:
                print("Saldo insuficiente")

        else:
            print("La tarjeta esta bloqueada")

    # Función para consultar saldo
    def consultar_saldo(self):
        print("Saldo disponible:", self.saldo_disponible)

    # Función para bloquear tarjeta
    def bloquear_tarjeta(self):
        self.estado_tarjeta = "Bloqueada"
        print("Tarjeta bloqueada")

    # Función para activar tarjeta
    def activar_tarjeta(self):
        self.estado_tarjeta = "Activa"
        print("Tarjeta activada")

    # Función para mostrar datos
    def mostrar_datos(self):

        print("\n----- DATOS DEL USUARIO -----")
        print("Usuario:", self.nombre_usuario)
        print("Numero de tarjeta:", self.numero_tarjeta)
        print("Estado:", self.estado_tarjeta)
        print("Limite de pago:", self.limite_pago)


# Crear objeto

usuario1 = Pago_Contactless(
    "1234-5678",
    5000,
    "Dylan",
    2000,
    "Activa"
)

# Menú

while True:

    print("\n===== PAGO CONTACTLESS =====")
    print("1. Mostrar datos")
    print("2. Consultar saldo")
    print("3. Realizar pago")
    print("4. Bloquear tarjeta")
    print("5. Activar tarjeta")
    print("6. Salir")

    opcion = input("Seleccione una opcion: ")

    if opcion == "1":
        usuario1.mostrar_datos()

    elif opcion == "2":
        usuario1.consultar_saldo()

    elif opcion == "3":
        usuario1.realizar_pago()

    elif opcion == "4":
        usuario1.bloquear_tarjeta()

    elif opcion == "5":
        usuario1.activar_tarjeta()

    elif opcion == "6":
        print("Programa finalizado")
        break

    else:
        print("Opcion no valida")
```
